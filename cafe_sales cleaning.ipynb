```python
net_sales = np.array(df["Net_Sales_INR"])
print(net_sales)
#converted net sales inr to numpy array
```

```python
#Calculated total sales using numpy 
total_sales = np.sum(net_sales)
print("TOTAL SALES:")
print(total_sales)
```

```python
# calculated average sales per order line
average_sales = np.mean(net_sales)
print("AVERAGE SALES PER ORDER LINE:")
print(average_sales.round(2))
```

```python
min_sales = np.min(net_sales)
max_sales = np.max(net_sales)
print("MINIMUM NET SALES:")
print(min_sales)
print("MAXIMUM NET SALES:")
print(max_sales)
#calculated maximum and minimum net sales
```

```python
median_sales = np.median(net_sales)
print("NET SALES MEDIAN:")
print(median_sales)
#calculation median of sales data
```

```python
std_deviation = np.std(net_sales)
print("STANDARD DEVIATION OF NET SALES:")
print(std_deviation.round(2))
#Calculated standard deviation
```

```python
#Calculate product_cost and profit
df["Total_Cost_INR"] = df["Quantity"] * df["Unit_Cost_INR"]
df["Profit_INR"] = df["Net_Sales_INR"] - df["Total_Cost_INR"]
print(df["Total_Cost_INR"])
print(df["Profit_INR"])
```

```python
#Calculating overall
total_revenue = df["Net_Sales_INR"].sum()
total_quantity = df["Quantity"].sum()
total_product = df["Total_Cost_INR"].sum()
total_profit = df["Profit_INR"].sum()
profit_margin_percentage = (total_profit/total_revenue)*100

print("TOTAL REVENUE : ₹" ,total_revenue)
print("TOTAL QUANTITY SOLD: ₹",total_quantity)
print("TOTAL PRODUCT COST: ₹", total_product)
print("TOTAL PROFIT: ₹", total_profit)
print(f"PROFIT MARGIN PERCENTAGE: {profit_margin_percentage:.2f}%\n")
```

```python
#total revenue and profit by each category
category_summary = (
    df.groupby("Category")
        .agg(
            Total_Revenue_INR = ('Net_Sales_INR','sum'),
            Total_Profit_INR = ('Profit_INR','sum')
         )
          .reset_index()
)
category_summary["Profit_Margin_%"] =(category_summary["Total_Profit_INR"]/ category_summary["Total_Revenue_INR"])*100
print('Category Breakdown:')
print(category_summary.round(2))
```

```python
top_5 = (
    df.groupby("Product",as_index= False)["Net_Sales_INR"]
      .sum()
      .nlargest(5 , "Net_Sales_INR")
      .reset_index(drop = True)
)
print("TOP FIVE PRODUCTS BASED ON REVENUE")
print(top_5)
#Calculate top 5 products based on revenue
```

```python
#groupby product and category

summary = df.groupby(["Product","Category"], as_index=False).agg(
    Total_units=('Quantity','sum'),
    Total_Revenue=('Net_Sales_INR','sum'),
    Total_profit=('Profit_INR','sum'),
    Avg_dscount_pct=('Discount_Percent','mean')
)
# calculate profit margin(%)    
summary['Profit_margin_%'] = (summary['Total_profit']/summary['Total_Revenue'])

#calculate simple avergae
avg_revenue = summary['Total_Revenue'].mean()
avg_margin = summary['Profit_margin_%'].mean()

#filter: above average revenue and below average profit margin

high_sales_low_profit = summary[
    (summary['Total_Revenue']> avg_revenue)&
    (summary['Profit_margin_%']< avg_margin)
    ].sort_values(by = 'Total_Revenue',ascending = False)

#display results and the ready  report line
print("=====HIGH SALES, LOW PROFIT ITEMS TABLE======")
print(high_sales_low_profit.round(2).to_string(index = False))
```
